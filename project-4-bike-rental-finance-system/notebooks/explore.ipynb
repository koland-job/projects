# Recalculation: look at the numbers

Read-only: the write plan is shown, but nothing is written to Google.
Writing is done by the "Fleet Ledger → Recalculate reports" menu item or `python -m fleet_ledger` from the project folder.
Run it from the project folder (next to `credentials/`); the sheet link comes from the `SPREADSHEET_URL` environment variable.

```python
from pathlib import Path

import pandas as pd

from fleet_ledger.plan import plan
from fleet_ledger.sheets import Spreadsheet

project_dir = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
snapshot = Spreadsheet(project_dir / "credentials").snapshot()
result = plan(snapshot)
for warning in result.warnings:
    print("⚠", warning)
```

```python
pd.DataFrame({name: pd.Series(result.reports[name]) for name in ["Free cash", "Cash flow", "P&L"]})
```
